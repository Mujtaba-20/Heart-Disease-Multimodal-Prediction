```python
x = df.iloc[:,:13]
y=df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.3)
```

```python
from sklearn.preprocessing import StandardScaler
scalar= StandardScaler()
x_train=scalar.fit_transform(x_train)
x_test=scalar.fit_transform(x_test)
```

```python
# Logistic Regression

from sklearn.linear_model import LogisticRegression
clf=LogisticRegression()
clf.fit(x_train,y_train)
```

```python
# RandomForest

from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
rf.fit(x_train,y_train)
```

```python
# XGBoost

from xgboost import XGBClassifier
xgb=XGBClassifier()
xgb.fit(x_train,y_train)
```

```python
# LightGBM

from lightgbm import LGBMClassifier
lgb=LGBMClassifier()
lgb.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
```

```python
# Logistic Regression Accuracy

clf_pred=clf.predict(x_test)
clf_acc=accuracy_score(y_test,clf_pred)*100
clf_acc
```

```python
# Random Forest Accuracy

rf_pred=rf.predict(x_test)
rf_acc=accuracy_score(y_test,rf_pred)*100
rf_acc
```

```python
# XGBoost Accuracy

xgb_pred=xgb.predict(x_test)
xgb_acc=accuracy_score(y_test,xgb_pred)*100
xgb_acc
```

```python
# LightGBM Accuracy

lgb_pred=lgb.predict(x_test)
lgb_acc=accuracy_score(y_test,lgb_pred)*100
lgb_acc
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score

class Heart_Disease:
    def __init__(self, clf, rf, xgb, lgb, feature_names, X_test_scaled, y_test_data, scaler):
        self.clf = clf
        self.rf = rf
        self.xgb = xgb
        self.lgb = lgb
        self.feature_names = feature_names
        
        self.X_test_scaled = X_test_scaled
        self.y_test_data = y_test_data
        self.scaler = scaler

        for i in range(1, 4): 
            print('='*50)
            print('Heart Disease Prediction')
            print('='*50)
            print('Main Menu Options')
            print('1. Enter the data through csv')
            print('2. Enter the data through input value')
            print('3. Exit')
            
            try:
                self.u_input = int(input('Enter the option no. to proceed: '))
            except ValueError:
                print('You have entered an invalid option format. Numbers only.\n')
                continue
            
            if self.u_input == 1:
                self.hcsv()
                break
            elif self.u_input == 2:
                self.htxt()
                break
            elif self.u_input == 3:
                self.exit_fxn()
                break
            else:
                print('You have entered an invalid option number\n')
        
    def hcsv(self):
        xyz = input('Enter the csv file with its extension: ')
        new_df = pd.read_csv(f'{xyz}').drop_duplicates()
        new_x = self.scaler.transform(new_df)

        clf_pred = self.clf.predict(new_x)
        rf_pred = self.rf.predict(new_x)
        xgb_pred = self.xgb.predict(new_x)
        lgb_pred = self.lgb.predict(new_x)

        clf_acc = accuracy_score(self.y_test_data, self.clf.predict(self.X_test_scaled)) * 100
        rf_acc = accuracy_score(self.y_test_data, self.rf.predict(self.X_test_scaled)) * 100
        xgb_acc = accuracy_score(self.y_test_data, self.xgb.predict(self.X_test_scaled)) * 100
        lgb_acc = accuracy_score(self.y_test_data, self.lgb.predict(self.X_test_scaled)) * 100

        if clf_acc >= rf_acc and clf_acc >= xgb_acc and clf_acc >= lgb_acc:
            print(clf_pred)
            print(f'Accuracy Score: {clf_acc:.2f}% ')
        elif rf_acc >= clf_acc and rf_acc >= xgb_acc and rf_acc >= lgb_acc:
            print(rf_pred)
            print(f'Accuracy Score: {rf_acc:.2f}% ')
        elif xgb_acc >= rf_acc and xgb_acc >= clf_acc and xgb_acc >= lgb_acc:
            print(xgb_pred)
            print(f'Accuracy Score: {xgb_acc:.2f}% ')
        elif lgb_acc >= rf_acc and lgb_acc >= clf_acc and lgb_acc >= xgb_acc:
            print(lgb_pred)
            print(f'Accuracy Score: {lgb_acc:.2f}% ')
        else:
            print('Error: Incorrect input.')

    def htxt(self):
        user_input = input('Enter the data to predict the result(separate values by comma): ')
        d = [float(x.strip()) for x in user_input.split(',')]
        data = pd.DataFrame([d], columns=self.feature_names)

        new_xt = self.scaler.transform(data)

        clf_pred = self.clf.predict(new_xt)
        rf_pred = self.rf.predict(new_xt)
        xgb_pred = self.xgb.predict(new_xt)
        lgb_pred = self.lgb.predict(new_xt)

        clf_acc = accuracy_score(self.y_test_data, self.clf.predict(self.X_test_scaled)) * 100
        rf_acc = accuracy_score(self.y_test_data, self.rf.predict(self.X_test_scaled)) * 100
        xgb_acc = accuracy_score(self.y_test_data, self.xgb.predict(self.X_test_scaled)) * 100
        lgb_acc = accuracy_score(self.y_test_data, self.lgb.predict(self.X_test_scaled)) * 100
        
        if clf_acc >= rf_acc and clf_acc >= xgb_acc and clf_acc >= lgb_acc:
            print(clf_pred)
            print(f'Accuracy Score: {clf_acc:.2f}% ')
        elif rf_acc >= clf_acc and rf_acc >= xgb_acc and rf_acc >= lgb_acc:
            print(rf_pred)
            print(f'Accuracy Score: {rf_acc:.2f}% ')
        elif xgb_acc >= rf_acc and xgb_acc >= clf_acc and xgb_acc >= lgb_acc:
            print(xgb_pred)
            print(f'Accuracy Score: {xgb_acc:.2f}% ')
        elif lgb_acc >= rf_acc and lgb_acc >= clf_acc and lgb_acc >= xgb_acc:
            print(lgb_pred)
            print(f'Accuracy Score: {lgb_acc:.2f}% ')
        else:
            print('Error: Incorrect input.')

    def exit_fxn(self):
        print('You have exited Successfully')
```
